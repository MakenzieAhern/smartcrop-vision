```python
import pandas as pd

# 1️⃣ Read four CSV files
agri = pd.read_csv('agriculture_dataset.csv')
recommend = pd.read_csv('Crop_recommendation.csv')
season = pd.read_csv('data_season.csv')
yield_data = pd.read_csv('crop_yield.csv')

# 2️⃣ Define the cleaning function
def clean_dataframe(df):
    df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
    df.drop_duplicates(inplace=True)
    df = df.dropna(axis=0, how='all')
    return df

# 3️⃣ Call function to clean data
agri_clean = clean_dataframe(agri)
recommend_clean = clean_dataframe(recommend)
season_clean = clean_dataframe(season)
yield_clean = clean_dataframe(yield_data)

# 4️⃣ Print Verification
print("Agriculture dataset shape:", agri_clean.shape)
print("Crop recommendation dataset shape:", recommend_clean.shape)
print("Season dataset shape:", season_clean.shape)
print("Crop yield dataset shape:", yield_clean.shape)
```

```python
agri_clean.to_csv('archive/agriculture_dataset_cleaned.csv', index=False)
recommend_clean.to_csv('archive/crop_recommendation_cleaned.csv', index=False)
season_clean.to_csv('archive/data_season_cleaned.csv', index=False)
yield_clean.to_csv('archive/crop_yield_cleaned.csv', index=False)
print("✅ All cleaned files saved to /archive/")
```
