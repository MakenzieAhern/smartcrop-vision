```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Setting display options
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)

# 1. Read all datasets
agri = pd.read_csv('agriculture_dataset.csv')
recommend = pd.read_csv('Crop_recommendation.csv')
season = pd.read_csv('data_season.csv')
yield_data = pd.read_csv('crop_yield.csv')

# 2. Print basic information of each dataset
datasets = {
    "Agriculture Dataset": agri,
    "Crop Recommendation": recommend,
    "Data Season": season,
    "Crop Yield": yield_data
}

for name, df in datasets.items():
    print(f"--- {name} ---")
    print(df.shape)
    print(df.columns)
    print(df.head(3), "\n")
```
